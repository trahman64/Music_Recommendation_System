## Checking missing values

```python
missing_values = music_df.isnull().sum()

print("Missing values:")
print(missing_values[missing_values > 0])

display(music_df[music_df.isnull().any(axis=1)])


print("Exact duplicate rows:")
print(music_df.duplicated().sum())

df_without_index = music_df.drop(columns=["Unnamed: 0"])

print("Duplicate rows excluding index:")
print(df_without_index.duplicated().sum())

print("Repeated track IDs:")
print(music_df["track_id"].duplicated().sum())


repeated_tracks = music_df[music_df.duplicated(subset="track_id", keep=False)]

display(repeated_tracks[["track_id", "track_name", "artists", "track_genre"]].head(20))


print("Tracks with zero duration:")
print((music_df["duration_ms"] == 0).sum())

print("Tracks with zero tempo:")
print((music_df["tempo"] == 0).sum())

print("Tracks with popularity outside 0–100:")
print((~music_df["popularity"].between(0, 100)).sum())


print("Genre distribution:")
print(music_df["track_genre"].value_counts().describe())


display(music_df[["duration_ms", "tempo", "loudness","energy", "danceability", "valence"]].describe())
```

```python
clean_df = music_df.copy()

clean_df = clean_df.drop(columns=["Unnamed: 0"])

clean_df = clean_df.dropna( subset=["track_name", "artists", "album_name"])

clean_df = clean_df.drop_duplicates()

clean_df = clean_df.reset_index(drop=True)

print("Original shape:", music_df.shape)
print("Cleaned shape:", clean_df.shape)
print("Remaining missing values:", clean_df.isnull().sum().sum())
print("Remaining duplicates:", clean_df.duplicated().sum())
selected_genres = [
    "pop", "rock", "classical", "hip-hop",
    "jazz", "electronic", "acoustic"
]

genre_df = clean_df[clean_df["track_genre"].isin(selected_genres)]

genre_summary = (
    genre_df.groupby("track_genre")[
        ["danceability", "energy"]
    ].mean().reset_index()
)

display(genre_summary)

fig = px.scatter(
    genre_summary,
    x="danceability",
    y="energy",
    color="track_genre",
    text="track_genre",
    title="Average Danceability vs. Energy Across Spotify Genres",
    labels={
        "danceability": "Average Danceability",
        "energy": "Average Energy",
        "track_genre": "Genre"
    }
)
fig
```

```python
selected_genres = [
    "pop", "rock", "classical", "hip-hop",
    "jazz", "electronic", "acoustic"
]

genre_df = clean_df[clean_df["track_genre"].isin(selected_genres)]

fig = px.box(
    genre_df,
    x="track_genre",
    y="danceability",
    color="track_genre",
    title="Distribution of Danceability Within Spotify Genres",
    labels={
        "track_genre": "Genre",
        "danceability": "Danceability"
    },
    points=False
)

fig.update_layout(showlegend=False)

fig.show()

genre_variation = (
    genre_df.groupby("track_genre")["danceability"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .round(3)
)

display(genre_variation)
```
